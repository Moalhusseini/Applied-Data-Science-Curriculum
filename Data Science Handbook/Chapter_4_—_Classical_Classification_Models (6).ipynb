# Theoretical Framework

## What is Classification?

Classification is a supervised machine learning task that aims to assign observations to predefined categories based on their characteristics. A classification model learns patterns from labeled training data and uses these learned patterns to predict the class of new, unseen observations.

Unlike regression models, which estimate continuous numerical values, classification models produce discrete outputs representing class membership. The predicted classes may consist of two categories (binary classification) or multiple categories (multiclass classification), depending on the nature of the problem.

The primary objective of classification is to accurately identify the class to which each observation belongs while minimizing classification errors. This enables organizations to automate decision-making processes and improve prediction accuracy across various domains.

Classification has become one of the most widely applied machine learning techniques because many real-world decision problems naturally involve categorical outcomes rather than continuous values.

## Types of Classification Problems

Classification problems can be categorized according to the number of target classes assigned to each observation. Selecting the appropriate classification approach depends on the nature of the prediction problem and the structure of the target variable.

The three most common types of classification are binary classification, multiclass classification, and multilabel classification.

### Binary Classification

Binary classification is the simplest and most widely used type of classification, where each observation belongs to one of two mutually exclusive classes. The objective is to determine which of the two possible outcomes best describes a given observation.

Binary classification problems are common in finance, healthcare, cybersecurity, and business intelligence. Examples include bankruptcy prediction, fraud detection, disease diagnosis, customer churn prediction, and spam email detection.

In this chapter, the Polish Companies Bankruptcy Dataset represents a binary classification problem, where companies are classified as either **Bankrupt** or **Non-Bankrupt**.

### Multiclass Classification

Multiclass classification extends binary classification by allowing each observation to belong to one of three or more distinct classes. Although multiple classes exist, each observation is assigned to only one class.

Examples include handwritten digit recognition, animal species classification, document categorization, and product classification.

Unlike binary and multiclass classification, multilabel classification allows an observation to belong to multiple classes simultaneously. Instead of predicting a single category, the model predicts a set of relevant labels.

Multilabel classification is frequently used in text categorization, image annotation, recommendation systems, and medical diagnosis, where multiple labels may correctly describe the same observation.

# Mathematical Foundations

Classification algorithms are built upon several mathematical concepts from probability theory, statistics, linear algebra, and optimization. These mathematical foundations provide the theoretical basis for learning decision boundaries, estimating class probabilities, measuring similarity between observations, and optimizing model parameters.

Although different classification algorithms employ different mathematical principles, they all aim to learn patterns from labeled data that allow accurate prediction of unseen observations. Understanding these mathematical concepts is therefore essential for interpreting model behavior and selecting appropriate algorithms for different classification problems.

The following sections introduce the fundamental mathematical concepts that will be used throughout this chapter.

## Probability Concepts

Probability is one of the fundamental concepts underlying supervised classification. Many classification algorithms estimate the likelihood that an observation belongs to a particular class before assigning the final prediction.

Mathematically, probability measures the chance that an event will occur and is expressed as a value between 0 and 1, where 0 indicates an impossible event and 1 indicates a certain event.

## Probability Concepts

Probability is a mathematical way of describing how likely an event is to occur. It is an important concept in classification because many classification models estimate the probability that an observation belongs to a particular class.

The value of a probability is always between 0 and 1:

$$
0 \leq P(A) \leq 1
$$

A probability close to 0 means that an event is unlikely to occur, while a probability close to 1 means that it is highly likely to occur.

For example, in the bankruptcy prediction problem, a model may estimate:

$$
P(\text{Bankrupt})=0.80
$$

This means that the model estimates an 80% probability that the company belongs to the bankrupt class.

Probability is particularly important in classification because it provides more information than the predicted class alone. Instead of only predicting whether a company is bankrupt or non-bankrupt, the model can also provide the estimated probability associated with the prediction.

### Observation

Probability provides the basic mathematical foundation for probabilistic classification. In this chapter, it is mainly used to understand how models such as Logistic Regression and Naïve Bayes estimate the likelihood of different classes.

## Conditional Probability

Conditional probability describes the probability of an event occurring when additional information about another event is already known.

It is important in classification because the probability of a class can change when new information about an observation becomes available.

The conditional probability of event \(A\) given event \(B\) is written as:

$$
P(A\mid B)
$$

This means the probability of \(A\) occurring given that \(B\) has occurred.

For example, in the bankruptcy prediction problem, let:

- \(A\) = the company is bankrupt
- \(B\) = the company has a high debt ratio

Then:

$$
P(\text{Bankrupt}\mid\text{High Debt})
$$

represents the probability that a company is bankrupt given that it has a high debt ratio.

The general formula is:

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
$$

where \(P(A\cap B)\) represents the probability that both events occur.

Conditional probability allows classification models to incorporate information about the observed features when estimating the probability of a particular class.

### Observation

Conditional probability measures how the probability of an event changes when additional information is available. It is an important foundation for probabilistic classification and is directly used in understanding **Bayes' Theorem** and **Naïve Bayes**.

## Bayes' Theorem

Bayes' Theorem is a fundamental concept in probability theory that explains how the probability of an event can be updated when new information becomes available.

In classification, Bayes' Theorem is used to estimate the probability that an observation belongs to a particular class based on its observed features.

The basic formula is:

$$
P(A\mid B)=
\frac{P(B\mid A)P(A)}
{P(B)}
$$

where:

- \(P(A\mid B)\) = probability of event \(A\) given \(B\)
- \(P(B\mid A)\) = probability of observing \(B\) given \(A\)
- \(P(A)\) = prior probability of event \(A\)
- \(P(B)\) = probability of observing \(B\)

For example, in bankruptcy prediction:

- \(A\) = company is bankrupt
- \(B\) = company has a high debt ratio

Then:

$$
P(\text{Bankrupt}\mid\text{High Debt})
$$

represents the probability that a company is bankrupt given that it has a high debt ratio.

Bayes' Theorem combines existing information about a class with new evidence from the observed features. This makes it particularly important for probabilistic classification methods such as **Naïve Bayes**.

### Observation

Bayes' Theorem provides a mathematical framework for updating class probabilities using observed evidence. In classification, it helps estimate how likely an observation is to belong to each possible class.

## Odds and Log-Odds

Odds describe the relationship between the probability that an event occurs and the probability that it does not occur.

They are particularly important in **Logistic Regression**, because the model uses log-odds to describe the relationship between the input features and the outcome.

The odds of an event can be calculated as:

$$
\text{Odds}(A)=\frac{P(A)}{1-P(A)}
$$

For example, if the probability that a company is bankrupt is 0.80, the odds are 4 to 1.

### Log-Odds

Log-odds, also called the **logit**, are the natural logarithm of the odds:

$$
\text{logit}(P)=\ln\left(\frac{P}{1-P}\right)
$$

Unlike probability, which ranges from 0 to 1, log-odds can take values from negative infinity to positive infinity.

In Logistic Regression, log-odds are modeled as a linear combination of the input features. This provides the connection between the model's features and its predicted probability.

### Observation

Odds compare the probability of an event with the probability that it does not occur. Log-odds transform this relationship into a scale that can be modeled using Logistic Regression.

## Sigmoid Function

The **sigmoid function** is a mathematical function that converts any real-valued input into a value between 0 and 1.

It is especially important in **Logistic Regression** because it allows the model to convert its output into a probability.

The sigmoid function is defined as:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

where \(z\) represents the input value and \(\sigma(z)\) represents the resulting probability.

For example:

$$
\sigma(0)=0.5
$$

As the input increases, the output approaches 1. As the input decreases, the output approaches 0.

The resulting probability can then be used to determine the predicted class using a decision threshold.

### Observation

The sigmoid function connects the output of Logistic Regression with probability-based classification. It transforms any real-valued input into a value between 0 and 1, making it suitable for representing the probability of the positive class.

## Decision Boundary

A **decision boundary** is the boundary that separates the regions associated with different predicted classes.

In binary classification, the model first estimates the probability of the positive class. A decision threshold is then used to convert this probability into a class label.

Using the standard threshold of 0.5:

$$
\hat{y}=
\begin{cases}
1, & P(Y=1\mid X)\geq0.5\\
0, & P(Y=1\mid X)<0.5
\end{cases}
$$

For example, in the bankruptcy prediction problem:

- Probability \(\geq 0.5\) → **Bankrupt**
- Probability \(< 0.5\) → **Non-Bankrupt**

The threshold does not always have to be 0.5. In practical applications, it can be adjusted depending on the objectives of the classification problem and the costs of false positive and false negative predictions.

### Observation

The decision boundary separates observations into different predicted classes. In Logistic Regression, the boundary depends on the learned model parameters and the selected probability threshold.

## Distance Metrics

Distance metrics are used to measure how similar or different two observations are. They are particularly important for **K-Nearest Neighbors (KNN)** because KNN uses distance to identify the observations that are closest to a new data point.

A smaller distance generally indicates greater similarity between observations.

### Common Distance Measures

One of the most common measures is **Euclidean distance**, which represents the straight-line distance between two observations.

Another common measure is **Manhattan distance**, which measures the total absolute difference between the feature values.

The choice of distance metric can affect which observations are considered neighbors and, consequently, the final classification.

### Feature Scaling

Distance-based algorithms are sensitive to the scale of the features. For example, a financial variable measured in millions could have a much larger influence on the distance than a ratio ranging between 0 and 1.

Therefore, numerical features should generally be scaled before applying KNN. Standardization is one common approach that places features on a comparable scale.

Importantly, the scaling parameters should be calculated using the training data and then applied to the validation and test data. This prevents information from the test set from influencing the training process.

### Observation

Distance metrics determine how KNN measures similarity between observations. Because the calculated distances directly influence the selected neighbors, choosing an appropriate distance measure and applying proper feature scaling are important steps when developing a KNN model.

## Margin and Margin Maximization

The **margin** is a central concept in Support Vector Machines (SVM). It represents the distance between the decision boundary and the closest observations from the different classes.

SVM aims to find a decision boundary that separates the classes while keeping the largest possible margin. The observations closest to the boundary are called **support vectors** because they have the greatest influence on the position of the decision boundary.

### Margin Maximization

A larger margin generally allows the model to create a more general decision boundary instead of focusing too closely on individual training observations.

The margin can be represented conceptually as:

$$
\text{Margin}=\frac{2}{\|w\|}
$$

where \(w\) represents the model's weight vector.

SVM therefore attempts to maximize the margin while maintaining an appropriate separation between the classes.

### Soft Margin

In real-world datasets, the classes are not always perfectly separable. SVM can therefore allow some observations to fall inside the margin or be incorrectly classified.

The parameter **\(C\)** controls the trade-off between a wider margin and classification errors.

A larger \(C\) places more emphasis on avoiding classification errors, while a smaller \(C\) allows more violations in exchange for a wider margin.

### Observation

Margin maximization is the main principle behind SVM. The model searches for a decision boundary that separates the classes while maintaining a suitable margin around it. The closest observations, known as support vectors, play an important role in determining the boundary, while \(C\) controls the balance between margin size and classification errors.

## Loss Functions and Optimization

Loss functions are used to measure how well a classification model's predictions match the actual class labels.

During training, the model tries to reduce this loss so that its predictions become more accurate.

### Common Classification Loss Functions

Different classification models can use different loss functions.

**Binary cross-entropy** is commonly associated with binary classification models such as Logistic Regression. It evaluates the quality of predicted probabilities, with lower loss indicating better probabilistic predictions.

**Hinge loss** is associated with Support Vector Machines (SVM). It focuses not only on whether observations are correctly classified, but also on whether they are classified with an appropriate margin.

### Optimization

Optimization is the process of finding model parameters that provide the lowest possible loss.

In simple terms, the learning algorithm repeatedly adjusts the model parameters to improve its predictions.

One common optimization approach is **gradient descent**, which updates the model parameters in the direction that reduces the loss.

### Regularization

Regularization is used to control model complexity and reduce the risk of overfitting.

It adds a penalty to the model objective when the model becomes unnecessarily complex. This encourages the model to learn patterns that generalize better to unseen data.

Regularization is particularly relevant for models such as Logistic Regression and SVM, where parameters can be controlled through regularization settings.

### Observation

Loss functions provide a way to measure prediction error, while optimization is used to improve the model by reducing this error. Different classification algorithms use different loss functions and optimization strategies. Regularization can further help control model complexity and improve generalization.

## Logistic Regression — Theory and Mathematical Model

Logistic Regression is a supervised classification algorithm used to estimate the probability that an observation belongs to a particular class.

Despite its name, Logistic Regression is mainly used for classification rather than predicting continuous numerical values.

In binary classification, the model estimates the probability of the positive class and then uses a decision threshold to assign the final class label.

### How Logistic Regression Works

Logistic Regression first combines the input features using learned coefficients:

$$
z=\beta_0+\beta_1x_1+\cdots+\beta_px_p
$$

The resulting value is then passed through the sigmoid function to obtain a probability between 0 and 1.

For example, in the bankruptcy prediction problem, the model can estimate the probability that a company is bankrupt based on its financial characteristics.

### Interpretation of Coefficients

Each feature has an associated coefficient that describes its relationship with the predicted outcome.

A positive coefficient means that increasing the feature is associated with higher log-odds of the positive class, while a negative coefficient indicates lower log-odds, assuming the other features remain constant.

Coefficients can also be converted into **odds ratios**:

$$
OR=e^\beta
$$

An odds ratio greater than 1 indicates higher odds of the positive class, while an odds ratio below 1 indicates lower odds.

This makes Logistic Regression useful not only for prediction but also for interpreting the relationship between financial variables and bankruptcy risk.

### Regularization

Regularization can be used to control model complexity and reduce overfitting.

In scikit-learn, Logistic Regression uses the parameter **\(C\)** to control the strength of regularization. Smaller values of \(C\) correspond to stronger regularization, while larger values allow the model to use larger coefficients.

The appropriate value of \(C\) can be selected through cross-validation during hyperparameter tuning.

### Classification Threshold

After estimating the probability of the positive class, the model converts the probability into a class label using a decision threshold.

The standard threshold is 0.5, but it can be adjusted depending on the objectives and error costs of the classification problem.

### Observation

Logistic Regression estimates class probabilities using a linear combination of the input features and the sigmoid function. Its coefficients provide useful information about the relationship between individual features and the predicted outcome. This combination of predictive ability and interpretability makes Logistic Regression a useful baseline model for financial distress prediction.

## K-Nearest Neighbors (KNN) — Theory and Mathematical Model

K-Nearest Neighbors (KNN) is a supervised classification algorithm that predicts the class of a new observation based on the classes of its closest training observations.

Unlike models such as Logistic Regression, KNN does not learn a fixed set of model coefficients during training. Instead, it stores the training observations and uses them when making a prediction.

The main idea is that observations that are close to each other in the feature space are likely to have similar class labels.

### How KNN Works

For a new observation, KNN identifies the \(k\) closest training observations using a selected distance metric.

The model then uses the classes of these neighbors to determine the predicted class. In standard KNN classification, the class with the largest number of votes is selected.

For example, if the five nearest neighbors contain three bankrupt companies and two non-bankrupt companies, the model predicts the new observation as **Bankrupt**.

### Choice of \(k\)

The value of \(k\) is an important hyperparameter.

A small value of \(k\) makes the model more sensitive to local patterns and individual observations. This can make the model more sensitive to noise.

A larger value of \(k\) considers more observations and produces a smoother decision structure. However, an excessively large value can make the model less sensitive to important local patterns.

Therefore, \(k\) should be selected using validation or cross-validation rather than chosen arbitrarily.

### Weighted KNN

In standard KNN, all selected neighbors have equal influence.

An alternative is **distance-weighted KNN**, where closer observations receive greater influence than observations that are farther away.

This can be useful when nearby observations are expected to provide more relevant information for the prediction.

### Computational Characteristics

KNN is often described as a **lazy learning** or **instance-based** algorithm because it performs little model fitting during training.

Most of the computational work occurs when predictions are made, as the algorithm needs to identify the nearest observations.

As the number of observations or features increases, prediction can become more computationally demanding.

### Feature Scaling

Feature scaling is particularly important for KNN because the model relies on distances.

If features have very different numerical ranges, variables with larger values can have a greater influence on the distance calculation.

Therefore, numerical features should generally be scaled before applying KNN. The scaling parameters must be learned from the training data and then applied to the validation and test data to avoid data leakage.

### Observation

KNN predicts the class of an observation based on its nearest neighbors. Its performance is strongly influenced by the choice of \(k\), the distance metric, and feature scaling. The model is simple and intuitive, but prediction can become computationally expensive for large datasets because distances must be evaluated when making predictions.

## Naïve Bayes — Theory and Mathematical Model

Naïve Bayes is a supervised classification algorithm based on probability and **Bayes' Theorem**.

The main idea is to calculate how likely an observation is to belong to each class and then choose the class with the highest probability.

### How Naïve Bayes Works

For a new observation, the model estimates the probability of each possible class based on the observed features.

The main equation is:

$$
P(C\mid X)=
\frac{P(X\mid C)P(C)}
{P(X)}
$$

where \(C\) represents the class and \(X\) represents the observed features.

The model then selects the class with the highest probability:

$$
\hat{C}=\arg\max_C P(C\mid X)
$$

### Conditional Independence Assumption

Naïve Bayes assumes that the features are **conditionally independent given the class**.

In simple terms, the model treats each feature as providing separate information about the class.

For example, when predicting bankruptcy, the model can consider financial indicators such as profitability, liquidity, and debt when estimating the probability of bankruptcy.

This assumption simplifies the calculation and makes the model computationally efficient. However, financial variables can be related to each other, so the assumption may not always hold perfectly in real-world data.

### Gaussian Naïve Bayes

Because the bankruptcy dataset contains continuous numerical financial variables, **Gaussian Naïve Bayes** can be used.

It assumes that the values of each numerical feature approximately follow a Gaussian distribution within each class.

The Gaussian distribution is represented by:

$$
P(x_j\mid C)=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{(x_j-\mu)^2}{2\sigma^2}
\right)
$$

where \(\mu\) represents the mean and \(\sigma^2\) represents the variance of the feature within the class.

### Advantages

- Simple and easy to understand
- Fast to train and predict
- Computationally efficient
- Produces probability estimates

### Limitations

The main limitation is the conditional independence assumption. When features are strongly related, the model's assumptions may not accurately represent the data.

### Observation

Naïve Bayes uses probability to determine the most likely class for an observation. Its simplicity and computational efficiency make it useful as a baseline classification model. For the bankruptcy dataset, Gaussian Naïve Bayes is suitable because the input variables are continuous financial indicators.

## K-Nearest Neighbors (KNN) — Theory and Mathematical Model

K-Nearest Neighbors (KNN) is a supervised classification algorithm that predicts the class of a new observation based on the classes of its closest training observations.

Unlike models such as Logistic Regression, KNN does not learn a fixed set of model coefficients during training. Instead, it stores the training observations and uses them when making a prediction.

The main idea is that observations that are close to each other in the feature space are likely to have similar class labels.

### How KNN Works

For a new observation, KNN identifies the \(k\) closest training observations using a selected distance metric.

The most common distance measure is **Euclidean distance**:

$$
d(x_i,x_j)=
\sqrt{\sum_{m=1}^{p}(x_{im}-x_{jm})^2}
$$

where \(p\) represents the number of features.

The model then uses the classes of the \(k\) nearest neighbors to determine the predicted class. In standard KNN classification, the class with the largest number of votes is selected.

For example, if the five nearest neighbors contain three bankrupt companies and two non-bankrupt companies, the model predicts the new observation as **Bankrupt**.

The predicted class can be expressed as:

$$
\hat{y}=
\operatorname{mode}(y_1,y_2,\ldots,y_k)
$$

where the \(y_i\) values represent the classes of the \(k\) nearest neighbors.

### Choice of \(k\)

The value of \(k\) is an important hyperparameter.

A small value of \(k\) makes the model more sensitive to local patterns and individual observations. This can make the model more sensitive to noise.

A larger value of \(k\) considers more observations and produces a smoother decision structure. However, an excessively large value can make the model less sensitive to important local patterns.

Therefore, \(k\) should be selected using validation or cross-validation rather than chosen arbitrarily.

### Weighted KNN

In standard KNN, all selected neighbors have equal influence.

An alternative is **distance-weighted KNN**, where closer observations receive greater influence than observations that are farther away.

This can be useful when nearby observations are expected to provide more relevant information for the prediction.

### Computational Characteristics

KNN is often described as a **lazy learning** or **instance-based** algorithm because it performs little model fitting during training.

Most of the computational work occurs when predictions are made, as the algorithm needs to identify the nearest observations.

As the number of observations or features increases, prediction can become more computationally demanding.

### Feature Scaling

Feature scaling is particularly important for KNN because the model relies on distances.

If features have very different numerical ranges, variables with larger values can have a greater influence on the distance calculation.

Therefore, numerical features should generally be scaled before applying KNN. The scaling parameters must be learned from the training data and then applied to the validation and test data to avoid data leakage.

### Observation

KNN predicts the class of an observation based on its nearest neighbors. Its performance is strongly influenced by the choice of \(k\), the distance metric, and feature scaling. The model is simple and intuitive, but prediction can become computationally expensive for large datasets because distances must be evaluated when making predictions.

## Support Vector Machine (SVM) — Theory and Mathematical Model

Support Vector Machine (SVM) is a supervised classification algorithm that finds a decision boundary that separates observations from different classes.

The main idea of SVM is to find a boundary that separates the classes while maximizing the distance between the boundary and the closest observations.

### How SVM Works

For a binary classification problem, the decision boundary can be represented as:

$$
w^Tx+b=0
$$

where \(w\) represents the model weights, \(x\) represents the input features, and \(b\) is the intercept.

The observations closest to the decision boundary are called **support vectors**. These observations have an important role in determining the position of the boundary.

### Margin

SVM aims to maximize the margin between the two classes.

The margin can be represented as:

$$
\text{Margin}=\frac{2}{\|w\|}
$$

A larger margin generally means that the decision boundary is farther from the closest observations.

### Soft Margin

In real-world datasets, the classes are not always perfectly separable. SVM can therefore allow some observations to violate the margin or be incorrectly classified.

The parameter **\(C\)** controls the trade-off between maximizing the margin and minimizing classification errors.

A larger \(C\) places more emphasis on avoiding classification errors, while a smaller \(C\) allows more violations in exchange for a wider margin.

### Kernel Function

SVM can also handle non-linear relationships using **kernel functions**.

A kernel allows the model to represent complex decision boundaries without explicitly transforming the original features into a higher-dimensional space.

Common kernel functions include:

- Linear
- Polynomial
- Radial Basis Function (RBF)

The RBF kernel is commonly used when the relationship between the features and classes is non-linear.

### Feature Scaling

Feature scaling is important for SVM because the algorithm is affected by the scale of the input features.

If features have very different numerical ranges, variables with larger values can have a greater influence on the model.

Therefore, numerical features should generally be scaled before training an SVM model. The scaling parameters must be learned from the training data and then applied to the validation and test data to avoid data leakage.

### Advantages

- Effective for high-dimensional data
- Can create linear and non-linear decision boundaries
- Uses support vectors to define the decision boundary
- Can perform well when the classes are clearly separated

### Limitations

- Sensitive to feature scaling
- Hyperparameters such as \(C\) and the kernel parameters need to be selected carefully
- Training can become computationally expensive for large datasets
- The resulting model can be less interpretable than Logistic Regression

### Observation

SVM classifies observations by finding a decision boundary that separates the classes while maximizing the margin. Support vectors determine the position of the boundary, while the parameter \(C\) controls the trade-off between margin size and classification errors. Kernel functions allow SVM to model non-linear relationships.

### Statistical Testing

The Mann–Whitney U test was used to compare the 15 selected financial indicators between bankrupt and non-bankrupt companies.

The Benjamini–Hochberg correction was applied because multiple financial indicators were tested simultaneously. This reduces the risk of false positive findings.

The t-test was not used because several financial indicators are highly skewed and contain extreme outliers, so the normality assumption required for the t-test is not appropriate.

```python
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests

results = []

for feature in important_features:
    g0 = df.loc[df["class"] == 0, feature].dropna()
    g1 = df.loc[df["class"] == 1, feature].dropna()
    results.append([feature, mannwhitneyu(g0, g1).pvalue])

statistical_results = pd.DataFrame(
    results,
    columns=["Feature", "P-Value"]
)

statistical_results["Adjusted P-Value"] = multipletests(
    statistical_results["P-Value"],
    method="fdr_bh"
)[1]

statistical_results.sort_values("Adjusted P-Value")
```

### Significant Financial Indicators

The indicators with an adjusted p-value below 0.05 are considered statistically significant.

```python
significant_features = statistical_results[
    statistical_results["Adjusted P-Value"] < 0.05
]

significant_features
```

### Feature Scaling

After completing missing-value treatment, redundancy reduction, and multicollinearity analysis, the dataset contains 25 final financial features.

Feature scaling is important because the financial indicators have very different numerical ranges. For example, some variables may contain small ratios, while others may have much larger values.

To place the features on a comparable scale, StandardScaler is used.
The scaler is fitted only on the training data and then applied to both the training and testing data. This prevents information from the testing set from being used during preprocessing.

Standardization
Standardization transforms each feature according to:
$$
z = \frac{x-\mu}{\sigma}
$$
where:
- x = original feature value
- μ = mean calculated from the training data
- σ = standard deviation calculated from the training data
- z = standardized value

After standardization, the training features should have approximately:
- Mean = 0
- Standard deviation = 1

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit the scaler on the training data
# and transform the training data
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train_vif_reduced),
    columns=X_train_vif_reduced.columns,
    index=X_train_vif_reduced.index
)

# Apply the same fitted scaler to the testing data
X_test_scaled = pd.DataFrame(
    scaler.transform(X_test_vif_reduced),
    columns=X_test_vif_reduced.columns,
    index=X_test_vif_reduced.index
)

print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)
```

### Cross-validation

Cross-validation is used to obtain a more reliable estimate of model performance and to reduce dependence on a single validation split.
Since the dataset is highly imbalanced, Stratified K-Fold Cross-Validation will be used in this study. Stratification ensures that each fold maintains approximately the same proportion of bankrupt and non-bankrupt companies.

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-Fold Stratified Cross-Validation configured successfully.")
```

###  Confusion Matrix for The Final Models

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

for ax, (name, model) in zip(axes, final_models.items()):

    y_pred = model.predict(X_test_vif_reduced)

    cm = confusion_matrix(y_test, y_pred)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Non-Bankrupt", "Bankrupt"],
        yticklabels=["Non-Bankrupt", "Bankrupt"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted Class")
    ax.set_ylabel("Actual Class")

plt.suptitle("Confusion Matrices - Final Models", fontsize=16)

plt.tight_layout()
plt.show()
```

### ROC Curves For the Models

```python
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

for ax, (name, model) in zip(axes, final_models.items()):

    y_prob = model.predict_proba(X_test_vif_reduced)[:, 1]

    fpr, tpr, _ = roc_curve(
        y_test,
        y_prob
    )

    roc_auc = auc(fpr, tpr)

    ax.plot(
        fpr,
        tpr,
        label=f"AUC = {roc_auc:.4f}"
    )

    ax.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Random Classifier"
    )

    ax.set_xlabel("False Positive Rate")
    ax.set_ylabel("True Positive Rate")
    ax.set_title(name)
    ax.legend()
    ax.grid(alpha=0.3)

plt.suptitle("ROC Curves - Final Models", fontsize=16)

plt.tight_layout()
plt.show()
```

## Results and Discussion

### 1. Descriptive Statistical Analysis and Exploratory Data Analysis

The descriptive and exploratory analysis showed substantial variability across the 64 financial indicators. Several variables had highly skewed distributions, large ranges, and extreme observations. In particular, Attr20, Attr43, Attr55, and Attr58 showed large differences between their mean and median values.

The dataset was highly imbalanced, containing 6,674 non-bankrupt and 271 bankrupt companies, corresponding to a bankruptcy rate of approximately 3.90%. Missing observations were mainly concentrated in Attr37 and Attr21.

The correlation analysis showed generally weak individual relationships with bankruptcy. Attr2 had the strongest positive correlation (0.0651), while Attr3 had the strongest negative correlation (-0.0851). The Mann–Whitney U test also indicated statistically significant differences between bankrupt and non-bankrupt companies for the 15 selected indicators.

Several financial indicators contained extreme observations and outliers. These observations were retained rather than automatically removed because they may represent genuine financial conditions and could contain useful information for bankruptcy prediction.

Overall, the EDA confirmed the presence of class imbalance, missing values, skewed distributions, extreme observations, and feature redundancy, supporting the need for appropriate preprocessing and feature selection.

### 2. Data Preprocessing

The preprocessing stage included duplicate removal, missing-value treatment, feature reduction, scaling, and class-imbalance handling.

A total of 82 duplicate records were removed, resulting in 6,945 observations. Attr37 was excluded because approximately 39% of its values were missing, while the remaining missing values were treated using median imputation.

The data were divided into 5,556 training observations and 1,389 test observations. Preprocessing parameters such as imputation and scaling were learned from the training data and then applied to the test data to avoid information leakage.

Because only approximately 3.9% of the companies were bankrupt, SMOTE was used within the training pipeline to balance the minority class. SMOTE was not applied to the test set, allowing the final evaluation to reflect the original class distribution.

### 3. Feature Selection

Feature selection was performed to reduce redundant variables and multicollinearity while retaining financially relevant predictors.

First, highly correlated features were identified using an absolute correlation threshold of 0.90. Representative variables were retained from highly correlated groups based on their relationship with the bankruptcy target. This reduced the feature set from 63 to 48 variables, followed by a second reduction from 48 to 29 variables.

VIF analysis was then applied to the remaining predictors. Additional variables with high multicollinearity and weaker target relationships were removed.

The final feature set contained 25 financial indicators:

Attr1, Attr3, Attr4, Attr5, Attr6, Attr7, Attr12, Attr15, Attr19, Attr21, Attr25, Attr27, Attr28, Attr29, Attr34, Attr41, Attr45, Attr47, Attr50, Attr55, Attr57, Attr59, Attr61, Attr63, Attr64.

The final VIF analysis showed substantially reduced multicollinearity. The highest remaining VIF was approximately 7.17 for Attr63, and no remaining feature pair had an absolute correlation of 0.90 or higher.

### 4. Model Development and Hyperparameter Tuning

Four classical classification models were evaluated: Logistic Regression, K-Nearest Neighbors (KNN), Naïve Bayes, and Support Vector Machine (SVM).

Five-fold stratified cross-validation was used for hyperparameter tuning. F1-score was selected as the optimization metric because the dataset was highly imbalanced and accuracy could be misleading.

| Model | Best Parameter | CV F1 |
|---|---|---:|
| Logistic Regression | C = 10 | 0.1156 |
| SVM | C = 10 | 0.1083 |
| KNN | n_neighbors = 15 | 0.0937 |
| Naïve Bayes | var_smoothing = 1e-9 | 0.0745 |

Logistic Regression achieved the highest cross-validation F1-score, making it the strongest candidate for final evaluation.

### 5. Final Model Evaluation

The tuned models were evaluated on the independent test set.

| Model | Accuracy | Precision | Recall | F1-Score | ROC-AUC |
|---|---:|---:|---:|---:|---:|
| Logistic Regression | 0.6609 | 0.0736 | 0.6667 | 0.1326 | 0.7185 |
| SVM | 0.6343 | 0.0684 | 0.6667 | 0.1241 | 0.7050 |
| KNN | 0.7171 | 0.0687 | 0.5000 | 0.1208 | 0.6403 |
| Naïve Bayes | 0.1073 | 0.0389 | 0.9259 | 0.0746 | 0.5299 |

The results demonstrate that accuracy alone is not an appropriate primary metric for this problem. Although KNN achieved the highest accuracy, its recall was only 0.5000, meaning that it failed to identify half of the bankrupt companies.

Naïve Bayes achieved the highest recall (0.9259), but its precision was only 0.0389, indicating a very large number of false-positive predictions.

Logistic Regression achieved the highest F1-score (0.1326) and ROC-AUC (0.7185), providing the best overall balance among the evaluated models.

### 6. ROC-AUC and Confusion Matrix Interpretation

The ROC-AUC results further supported Logistic Regression as the strongest model:

- Logistic Regression: 0.7185
- SVM: 0.7050
- KNN: 0.6403
- Naïve Bayes: 0.5299

The confusion matrices showed that Logistic Regression correctly identified 36 of the 54 bankrupt companies in the test set, while missing 18. It also generated false-positive predictions, demonstrating the trade-off between increasing bankruptcy detection and controlling false alarms.

Naïve Bayes identified 50 of the 54 bankrupt companies, but at the cost of a very high number of false positives. This explains why its high recall did not translate into a high F1-score.

### 7. Final Model Selection

Based on the combined cross-validation and independent test results, Logistic Regression was selected as the final model.

The selection was based primarily on F1-score and ROC-AUC rather than accuracy, because the dataset contains a severe class imbalance. Logistic Regression achieved:

- CV F1-score: 0.1156
- Test F1-score: 0.1326
- Test ROC-AUC: 0.7185
- Test Recall: 0.6667

These results indicate that Logistic Regression provided the strongest overall performance among the four classical classification approaches.

### 8. Overall Discussion

Overall, the results show that bankruptcy prediction is a challenging problem due to the strong class imbalance, skewed financial variables, extreme observations, and complex relationships among financial indicators.

The preprocessing and feature selection stages reduced the original predictors to 25 less redundant financial indicators. Among the evaluated models, Logistic Regression provided the best overall balance between identifying bankrupt companies and limiting false-positive predictions.

However, the relatively low precision and F1-score indicate that the model still has considerable room for improvement. Therefore, Logistic Regression should be considered a strong classical baseline rather than a fully optimized standalone bankruptcy prediction system.

## Conclusion

This chapter presented a classical machine learning framework for bankruptcy prediction using the Polish Companies Bankruptcy Dataset.

The exploratory analysis identified severe class imbalance, missing values, skewed financial indicators, extreme observations, and feature redundancy.

After preprocessing and feature selection using correlation analysis and VIF, the number of predictors was reduced to 25 financial indicators. SMOTE was applied during training to improve minority-class learning, while the independent test set retained its original distribution.

Four classification models were evaluated: Logistic Regression, KNN, Naïve Bayes, and SVM. Five-fold stratified cross-validation was used for hyperparameter tuning, with F1-score as the main optimization metric.

Logistic Regression achieved the best overall performance, with a test F1-score of 0.1326 and ROC-AUC of 0.7185, while correctly identifying 36 of 54 bankrupt companies. Although Naïve Bayes achieved the highest recall (0.9259), its very low precision resulted in a lower F1-score.

Overall, Logistic Regression was selected as the strongest classical baseline. However, its relatively low F1-score and precision show that bankruptcy prediction remains challenging and provides motivation for evaluating more advanced models in subsequent chapters.
